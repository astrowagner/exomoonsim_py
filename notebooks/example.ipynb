# exomoonsim — example

Reproduces a single detection trial and a small survey figure. See `README.md` for details.

```python
import numpy as np
import matplotlib.pyplot as plt
from exomoonsim.sim import SimParams, run_trial
from exomoonsim.survey import SurveyConfig, run_survey, SurveyResult
from exomoonsim.plots import plot_survey
```

## 1. A single trial
An Earth-mass moon at 10 R$_{\rm Jup}$, observed at 10 µas precision.

```python
p = SimParams(moon_a=10.0, moon_mass=1.0, astrometric_precision=1e-5)
r = run_trial(p, seed=1)
for k in ['sig', 'perr', 'amperr', 'best_period', 'sidereal_period', 'moon_period',
          'input_amp', 'retrieved_amp', 'fit_success']:
    print(f'{k:18s} {r[k]}')
```

## 2. A small survey + figure
Fast 3×3 grid. Bump `ntrials` and switch to the full grid (drop `a_grid`/`mass_grid`) for science runs.

```python
cfg = SurveyConfig(a_grid=np.array([5., 10., 20.]),
                   mass_grid=np.array([0.1, 1.0, 5.0]),
                   ntrials=6,
                   base=SimParams(astrometric_precision=1e-5))
res = run_survey(cfg)
res.save('fast_survey.npz')
fig = plot_survey(res, filename='fast_survey.png')
plt.show()
```

## 3. Re-plot a saved run
`SurveyResult.load` reads a `.npz` written by the CLI or `run_survey`, so you can restyle the figure without recomputing.

```python
res2 = SurveyResult.load('fast_survey.npz')
plot_survey(res2)
plt.show()
```
